```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace (0, 10, 40) #Genera un array con 40 valores desde el 0 al 10
y = np.sin (x)
```

```python
plt.plot(x,y, "o")
```

```python
ran = np.random.RandomState(0)
marcadores = ["o", "x", "+", "v", "s", "d"]
```

```python
for marcador in marcadores:
    plt.plot(ran.rand(5), ran.rand(5), marcador)
```

```python
plt.plot(x,y, ":pb"); #los ":" son para lo que hay entre los puntos, si queremos linea, vacío o lineas discontinuas, la "p" el tipo de punto y la "b" el color
```

```python
plt.scatter(ran.randn(100), ran.randn(100),
            c = ran.randn(100),
            s = ran.randn(100) * 1000,
            alpha = 0.75,)
plt.colorbar();
```

```python
t = np.linspace(0, 2*np.pi, 100)
x1 = 16 * np.sin(t)**3
y1 = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)

plt.figure(figsize=(8,6))
plt.plot(x1, y1, color='red')
plt.fill(x1, y1, color='pink')
plt.axis('equal')
plt.grid();
```
