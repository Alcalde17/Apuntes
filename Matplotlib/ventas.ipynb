```python
numeros = np.random.randn(1000) #Genera 1000 numeros con una distribución gausiana nacida desde 0 con una desviación estándar de 1, es decir la mayoría están entre el 1 y el -1
plt.hist(numeros);
```

```python
plt.hist(numeros, bins= 60, alpha = 0.5, color = "grey", edgecolor= "white");
#Bins es para hacer los muestreos más precisos
#Alpha es la opacidad y edgecolor es el color de los bordes del gráfico
```

```python
plt.hist(numeros, bins=40, histtype = "step"); #histtype es para darle un formato diferente al gráfico
```

```python
n1 = np.random.randn(1000)
n2 = np.random.randn(500)
n3 = np.random.randn(200)

plt.hist(n1, bins=40, alpha=0.3, color="red")
plt.hist(n2, bins=40, alpha=0.3, color="yellow")
plt.hist(n3, bins=40, alpha=0.3, color="blue");
```
