```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/ruben/Desktop/ESTUDIOS/Curso Analisis de datos/ProyectoInicial/Matplotlib/Datos+Meteorológicos_Arg_2023.csv")
df
```

```python
df["Fecha"] = pd.to_datetime(df["Fecha"], format = "%d/%m/%Y")
df["Mes"] = df["Fecha"].dt.month
df
```

```python
def consulta():
    lista_ciudades = ["Buenos Aires","Cordoba","Rosario","Mendoza","Tucuman"]
    while True:
        ciudad = input("Seleccione la ciudad: (Buenos Aires / Cordoba / Rosario / Mendoza / Tucuman): ")
        mes = int(input("Seleccione el mes: (1-12): "))

        if ciudad not in lista_ciudades or mes not in range(1,13):
            print("Seleccione una ciudad o mes válidos")
            continue

        df_nuevo = df[(df["Ciudad"] == ciudad) & (df["Mes"] == mes)]

        plt.figure(figsize=(10,6))
        plt.plot(df_nuevo["Fecha"], df_nuevo["Temperatura Maxima"], label="Maxima", color = "red")
        plt.plot(df_nuevo["Fecha"], df_nuevo["Temperatura Minima"], label="Minima", color = "blue")
        plt.title(f"Temperaturas en {ciudad} en el mes {mes}")
        plt.xlabel("Fecha")
        plt.ylabel("Temperatura")
        plt.legend()
        plt.xticks(rotation = 45);

        otra_consulta = input("Deseas continuar consultando (s/N): ")
        if otra_consulta.lower().strip() != "s":
            break
```

```python
consulta()
```
