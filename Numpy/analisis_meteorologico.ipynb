## Limpiamos los datos

```python
filtro = df["Temperatura"].isna() & df["Precipitación"].isna() & df["Humedad"].isna()
df = df.drop(df[filtro].index)
df.info()
```

## Pasar el DataFrame a arrays de Numpy

```python
temperatura = df["Temperatura"].to_numpy()
precipitacion = df["Precipitación"].to_numpy()
humedad = df["Humedad"].to_numpy()
```

### Funcion para hacerlo en los 3 arrays

```python
def remplazar_por_media(array):
    filtro = np.isnan(array)
    media = np.nanmean(array)
    return np.where(filtro, media, array)
```

```python
temperatura = remplazar_por_media(temperatura)
precipitacion = remplazar_por_media(precipitacion)
humedad = remplazar_por_media(humedad)
```

## Analisis rapido de datos
#### Temperatura promedio

```python
temp_prom = np.mean(temperatura)
temp_prom
```

#### Máxima humedad registrada

```python
hum_max = np.max(humedad)
hum_max
```
