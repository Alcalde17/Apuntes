```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
```

# CLUSTERING JERÁRQUICO (Hierarchical Clustering)
------------------------------------------------
Este tipo de clustering es un método de aprendizaje no supervisado que sirve para<br>
agrupar datos en función de su similitud sin necesidad de indicar previamente<br>
cuántos grupos (clusters) queremos.<br><br>

La idea principal es construir una jerarquía de clusters:<br>
 - Al inicio, cada punto es su propio cluster<br>
 - En cada paso, se unen los dos clusters más cercanos<br>
 - Se repite hasta que todos los puntos forman un único cluster<br><br>

 Esto permite visualizar la estructura de los datos y decidir después cuántos grupos tiene sentido formar.


```python
iris = load_iris()
```

```python
X = iris.data
```

### CONSTRUCCIÓN DEL CLUSTERING JERÁRQUICO
 ------------------------------------------------
 linkage:<br>
 - Construye la jerarquía de clusters<br>
 - Devuelve información sobre cómo se van fusionando los grupos<br><br>

 Método "ward":<br>
 - Une clusters minimizando la varianza interna<br>
  - La opcion mas usada normalmente<br>
 - Genera grupos compactos y bien definidos<br><br>
##### Tipos de linkage
single   → detecta formas raras pero puede encadenar<br>
complete → clusters compactos, sensible a outliers<br>
average  → equilibrio general<br>
centroid → basado en medias, menos usado<br>
ward     → minimiza varianza, muy usado en práctica

```python
linked = linkage(X, "ward")
```

### VISUALIZACIÓN: DENDROGRAMA
 ------------------------------------------------

 El dendrograma es un gráfico en forma de árbol que muestra:<br>
 - Cómo se agrupan los datos paso a paso<br>
 - A qué distancia se unen los clusters<br><br>

 Interpretación:<br>
 - Uniones bajas → elementos muy similares<br>
 - Uniones altas → clusters diferentes<br><br>

 Sirve especialmente para:<br>
 - Decidir el número óptimo de clusters (cortando el árbol horizontalmente)<br>
 - Entender la estructura de los datos<br>
 - Detectar outliers (puntos que se unen muy tarde)


```python
plt.figure(figsize=(10,7))
dendrogram(linked,
          orientation = "top",
          labels = iris.target,
          distance_sort="descending",
          show_leaf_counts=True)
plt.title("Dendograma de clustering jerarquico");
```
