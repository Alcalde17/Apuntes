```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import seaborn as sns
```

```python
np.random.seed(42)
data = np.random.rand(100,2)
data
```

```python
kmeans = KMeans(n_clusters = 5, random_state = 42)
kmeans.fit(data)
```

```python
centroides = kmeans.cluster_centers_
centroides
```

```python
etiquetas = kmeans.labels_
etiquetas
```

```python
plt.figure(figsize=(10,6))
plt.scatter(data[:,0], data[:,1], c = etiquetas, cmap="Reds", marker = "o");
```

```python
plt.figure(figsize=(10,6))
plt.scatter(data[:,0], data[:,1], c = etiquetas, cmap="Reds", marker = "o")

plt.scatter(centroides[:,0], centroides[:,1], marker = "x", s = 100, c="blue", label="Centroides");
plt.legend()
```

```python
pinguinos = sns.load_dataset("penguins")
pinguinos.dropna(inplace=True)
data = pinguinos[["bill_length_mm", "bill_depth_mm"]]
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(data)
```

```python
centroides = kmeans.cluster_centers_
centroides
```

```python
etiquetas = kmeans.labels_ #Agrupa cada punto con el centroide que tenga mas cerca
etiquetas
```

```python
plt.figure(figsize=(12,7))
sns.scatterplot(data=pinguinos,
                x="bill_length_mm",
                y="bill_depth_mm",
                hue=etiquetas,
                palette="colorblind")
plt.scatter(centroides[:,0], centroides[:,1], c="red", s=150, marker="x", label="Centroides")
plt.title("Kmeans clustering en el conjunto de datos de pinguinos")
plt.legend()
plt.xlabel("Longuitud del pico en mm")
plt.ylabel("Ancho del pico en mm");
```
