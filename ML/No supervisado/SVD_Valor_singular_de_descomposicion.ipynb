```python
import numpy as np
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
```

# SVD Valor singular de descposicion
Sirve para, al igual que el PCA reducir la dimensionalidad y encontrar las dimensiones principales que agrupan a los datos en nuestro proyecto

```python
A = np.array([[1,2],
              [3,4],
              [5,6]])
```

```python
U, sigma, VT = np.linalg.svd(A)
```

```python
U #La variable U nos dice donde esta cada dato en los nuevos ejes
```

```python
sigma #Contiene la importancia de la información de los valores principales
```

```python
VT #Dice la dirección de los nuevos ejes
```

```python
iris = load_iris()
X = iris.data
```

```python
x_centrado = X -np.mean(X, axis = 0) #Normalización de datos de X (iris).
```

```python
U,sigma, VT = np.linalg.svd(x_centrado) #Lo descomponemos en 3 valores. La variable sigma contiene la importancia de los componentes en el dataset
```

```python
k = 2 #Establecemos que sea un espacio bidimensional
x_transformado = U[:, :k] * sigma[:k] # Selecciona las 2 primeras columnas (k) y escala cada valor de U por la importancia que tienen
```

```python
especies = ["setosa", "versicolor", "virginica"]
```

```python
plt.figure(figsize = (10,8))
for i in range(3):
    plt.scatter(x_transformado[iris.target == i, 0],
                x_transformado[iris.target == i, 1],
                label = especies[i])
plt.xlabel("Primer componente principal")
plt.ylabel("Segundo componente principal")
plt.legend()
plt.title("Dataset Iris transformado por SVD");
```
