```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
dimensiones = (4,4)
estado_inicial = (0,0)
estado_objetivo = (3,3)
acciones = [(0,-1),
            (0,1),
            (-1,0),
            (1,0)]
acciones_simbolos=["↑","↓","←","→"]
```

```python
num_estados = dimensiones[0] * dimensiones[1]
num_estados
```

```python
num_acciones = len(acciones)
num_acciones
```

```python
Q = np.zeros((num_estados, num_acciones))
print(Q)
```

```python
alpha = 0.1
gamma = 0.99
epsilon = 0.2
episodios = 1000
```

```python
def estado_a_indice(estado):
    return estado[0] * dimensiones[1] + estado[1]
```

```python
estado_a_indice((3,1))
```

```python
def elegir_accion(estado):
    if random.uniform(0,1) < epsilon:
        return random.randint(0,num_acciones-1) #Exploración
    else:
        return np.argmax(Q[estado_a_indice(estado)])#Encontrar la mejor accion
```

```python
def aplicar_accion(estado, accion_idx):
    accion = acciones[accion_idx]
    nuevo_estado = tuple(np.add(estado, accion) % np.array(dimensiones))
    # % np.array(dimensiones hace que el escenario se vuelva un espacio toroidal para atravesar limites y aparecer en el otro lado
    if nuevo_estado == estado_objetivo:
        recompensa = 1
    else:
        recompensa = -1
    return nuevo_estado, recompensa, nuevo_estado == estado_objetivo
```

```python
for episodio in range(episodios):
    estado = estado_inicial
    accion_idx = elegir_accion(estado)
    terminado = False

    while not terminado:
        nuevo_estado, recompensa, terminado = aplicar_accion(estado, accion_idx)
        nueva_accion_idx = elegir_accion(nuevo_estado)
        indice = estado_a_indice(estado)
        Q[indice, accion_idx] += alpha * (recompensa + gamma * Q[estado_a_indice(nuevo_estado), nueva_accion_idx] - Q[indice, accion_idx]) #Algoritmo de SARSA
        estado,accion_idx = nuevo_estado, nueva_accion_idx
```

```python
politica_simbolos = np.empty(dimensiones, dtype="<U2") #Estotodo es necesario para hacer un array vacio con las dimensiones de dimensiones
politica_simbolos
```

```python
for i in range(dimensiones[0]):
    for j in range(dimensiones[1]):
        estado = (i,j)
        mejor_accion = np.argmax(Q[estado_a_indice(estado)])
        politica_simbolos[i,j] = acciones_simbolos[mejor_accion]
print(politica_simbolos)
```
