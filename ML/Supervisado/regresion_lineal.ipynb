```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.DataFrame()
df["Area"]=[2600,3000,3200,3600,4000]
df["Precio"]=[550,565,610,680,725]
df
```

```python
plt.scatter(df["Area"],
            df["Precio"]);
```

```python
X = df[["Area"]]
```

```python
y = df["Precio"]
```

```python
modelo = linear_model.LinearRegression()
```

```python
modelo.fit(X,y)
```

```python
modelo.predict([[685]])
```

```python
plt.plot(df["Area"],
         modelo.predict(X));
```

```python
plt.scatter(df["Area"],
            df["Precio"])
plt.plot(df["Area"],
         modelo.predict(X));
```
