# Bosques Aleatorios
Los bosques aleatorios funcionan construyendo varios arboles de decision, comparando y combinando sus resultados para hacer una prediccion final mejorando así su precisión.<br>
Son menos propensos al sobreajuste en comparacion a los arboles de decisión.<br>
Estos se pueden usar paraproblemas tanto de regresion como de clasificacion, haciendolos versatiles para diferentes tipos de datos y analisis.<br><br>

En este Dataset usaremos la columna "Clase" para predecir predicciones, basandonos en relaccione scon otras variables.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

from sklearn.preprocessing import MinMaxScaler
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv("datos_tarjetas/tarjetas_credito.csv")
df
```

### Normalizacion de datos
Normalizar los datos para que el el valor min sea 0 y el max 1 para que nuestro agente vea relacciones más facilmente

```python
escala = MinMaxScaler(feature_range=(0,1))
normado = escala.fit_transform(df)
df_normado = pd.DataFrame (data=normado, columns = df.columns)
df_normado
```

```python
X = df_normado.drop("Clase", axis=1) # La X es todas las columnas menos clase
```

```python
y = df_normado["Clase"] # La y es la columna clase
```

### Entrenamos los datos

```python
X_entrena, X_prueba, y_entrena, y_prueba = train_test_split(X,y,train_size=0.7, random_state=42)
```

```python
forest= RandomForestClassifier()
forest.fit(X_entrena, y_entrena)
```

```python
forest.score(X,y)
```
