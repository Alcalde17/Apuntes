```python
import seaborn as sns
```

```python
propinas = sns.load_dataset("tips")
propinas
```

```python
sns.catplot(data=propinas,
            kind="strip",
            x="day",
            y="total_bill",
            hue="smoker");
```

```python
sns.catplot(data=propinas,
            kind="swarm", #separa los puntos para que sea mas visible cada punto
            x="day",
            y="total_bill",
            hue="smoker");
```

```python
sns.catplot(data=propinas,
            kind="boxen", #es un boxplot la caja recoge los datos del 75% al 25% y te da los outliers
            x="day",
            y="total_bill",
            hue="smoker");
```

```python
sns.catplot(data=propinas,
            kind="violin", #el grosor indica la densidad de datos
            x="day",
            y="total_bill",
            hue="smoker");
```

```python
sns.catplot(data=propinas,
            kind="point", #cada punto indica la media y las lineas verticales son la variabilidad
            x="day",
            y="total_bill",
            hue="smoker");
```

```python
sns.catplot(data=propinas,
            kind="bar", #separa los puntos para que sea mas visible cada punto y las lineas son los intervalos de confianza
            x="day",
            y="total_bill",
            hue="smoker");
```
