```python
import seaborn as sns
```

```python
diamantes = sns.load_dataset("diamonds")
diamantes
```

```python
sns.displot(data=diamantes,
            x = "carat",
            y = "price",
            hue="clarity",
            col = "cut");
```

```python
sns.catplot(data = diamantes,
            x = "y",
            y = "price",
            hue = "color",
            col = "clarity")
sns.set_palette("colorblind");
```

```python
sns.catplot(data = diamantes,
            y = "table",
            x = "clarity",
            col = "color",
            kind = "box")
```

```python
sns.jointplot(data = diamantes,
            x = "carat",
            y = "x",
            hue = "cut");
```

```python
sns.pairplot(data=diamantes,
             hue = "cut")
```

```python
sns.displot(data = diamantes,
            x = "x",
            y= "table",
            kind="kde",
            col="cut",
            fill = True,
            color = "blue");
```
