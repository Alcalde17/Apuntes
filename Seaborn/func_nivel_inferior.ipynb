```python
import seaborn as sns
pinguinos = sns.load_dataset("penguins")
```

```python
g = sns.PairGrid(pinguinos,
                 hue="species")
#Nos crea gráficos vacíos pero si te fijas se duplican gráficos
```

```python
g = sns.PairGrid(pinguinos,
                 hue="species",
                 corner = True)
#Así nos quitamos los gráficos duplicados simétricamente
```

```python
g = sns.PairGrid(pinguinos,
                 hue="species",
                 corner = True)
g.map_lower(sns.kdeplot,        #Al trabajar con bajo nivel tenemos que poner el nombre del grafico que queramos +"plot"
            hue = None,
            levels = 11,
            color = "0.2");
```

```python
g = sns.PairGrid(pinguinos,         #Lienzo en blanco
                 hue="species",
                 corner = True)
g.map_lower(sns.kdeplot,            #Analisis de densidad topográfico global
            hue = None,
            levels = 11,
            color = "0.2");
g.map_lower(sns.scatterplot,        #Analisis de distribución individual por especies
            marker="+")
```

```python
g = sns.PairGrid(pinguinos,         #Lienzo en blanco
                 hue="species",
                 corner = True)
g.map_lower(sns.kdeplot,            #Analisis de densidad topográfico global
            hue = None,
            levels = 9,
            color = "0.2");
g.map_lower(sns.scatterplot,        #Analisis de distribución individual por especies
            marker="+")
g.map_diag(sns.histplot,            #Hacemos un histograma a los graficos centrales, los diagonales
           element="step",
           linewidth=0,
           kde=True)
g.add_legend(frameon=True)          #mostramos la leyenda y la ubicamos donde queremos
g.legend.set_bbox_to_anchor((.61,.6))
```
