```python
import seaborn as sns
```

```python
propinas = sns.load_dataset("tips")
```

```python
propinas
```

```python
sns.relplot(data = propinas,
            x = "total_bill",
            y = "tip",
            col = "sex",
            style = "time");
```

```python
sns.relplot(data = propinas,
           x = "total_bill",
           y = "tip")
```

# Seaborn
Vamos a ver que significa cada linea que tenemos aqui debajo:<br>
<b>data</b>: el dataframe que queremos usar para mostrar los datos<br>
<b>x / y</b>: las columnas del dataframe que queremos usar como eje x/y <br>
<b>hue</b>: separar por color <br>
<b>style</b>: separar por forma <br>
<b>size</b>: hacer mas grande o mas pequeño <br>
<b>col</b>: nos separa por graficos(columnas) <br>
<br>
Todos estos elementos usan una columna del dataframe como referencia a hacer sus acciones

```python
sns.relplot(data = propinas,
           x = "total_bill",
           y = "tip",
            hue = "smoker",
            style = "smoker",
            size = "size",
            col = "time");
```
