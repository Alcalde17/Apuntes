```python
import seaborn as sns
pinguinos = sns.load_dataset("penguins")
```

```python
sns.jointplot(data=pinguinos,
              x="flipper_length_mm",
              y="bill_length_mm",
              hue="species");
#Nos da la distribución conjunta y las marginales
```

```python
sns.pairplot(data=pinguinos,
             hue="species");
#Compara toda las variables numéricas del dataset y crea un grid de gráficos lo que permite evaluar rápidamente y comprender posibles relaciones
```
