```python
import seaborn as sns
```

```python
pinguinos = sns.load_dataset("penguins")
pinguinos
```

```python
sns.relplot(data=pinguinos,
           x="bill_length_mm",
           y="bill_depth_mm",
           hue="body_mass_g"
           );
```

### Personalizacion:
Para personalizar nuestro grafico tenemos que guardarlo en una variable para poder hacer las modificaciones pertinentes

```python
sns.set_theme(style="white")
g = sns.relplot(data=pinguinos,
           x="bill_length_mm",
           y="bill_depth_mm",
           size="body_mass_g",
           marker="^",
            hue="species");
```

```python
g = sns.relplot(data=pinguinos,
           x="bill_length_mm",
           y="bill_depth_mm",
           size="body_mass_g",
           marker="^",
            hue="species")

g.set_axis_labels("Longitud del pico (mm)",     #Esto nos permite que nuestro grafico no tenga el nombre como tal de la serie de nuestro DF, nos permite darle un estilo y un nombre
                 "Grosor del pico (mm)",
                 labelpad=10)
g.legend.set_title("Especie y Masa(g)")         #Aqui nos centramos en la leyenda, por lo que llamamos a la leyenda y de ahi por ejemplo le damos un titulo más ilustrativo
;
```

```python
g = sns.relplot(data=pinguinos,
           x="bill_length_mm",
           y="bill_depth_mm",
           size="body_mass_g",
           marker="^",
            hue="species")

g.set_axis_labels("Longitud del pico (mm)",
                 "Grosor del pico (mm)",
                 labelpad=10,
                  fontsize= 17)
g.legend.set_title("Especie y Masa(g)")
g.figure.set_size_inches(14.5,11.5)                     #Podemos tambien modificar el grafico con figure para, por ejemplo, cambiarle el tamaño para ver mas precisamente la dispersion de datos
g.figure.suptitle("Tamaño del pico en pingüinos",       #Ademas podríamos darle un título, cambiando el tipo de letra y el tamaño acorde con el del gráfico
                  fontsize = 20,
                  fontweight="bold");
```
