```python
import seaborn as sns
```

```python
pinguinos = sns.load_dataset("penguins")
pinguinos
```

```python
sns.displot(data=pinguinos,
            x="body_mass_g");
```

```python
sns.displot(data=pinguinos,
            x="body_mass_g",
            kde=True, #Traza una linea de densidad probable
            col="island");
```

```python
sns.displot(data=pinguinos,
            x="flipper_length_mm",
            kind="kde",
            rug=True);
```

```python
sns.displot(data=pinguinos,
            y="body_mass_g",
            x="sex",
            color="gray")
```

```python
sns.displot(data=pinguinos,
            y= "body_mass_g",
            kind="ecdf"); #nos dice el nº de pinguinos que pesa eso o menos, y nos sirve para ver como van distribuidos
```
