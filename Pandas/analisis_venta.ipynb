# Analisis de ventas
## Lectura de datos:

```python
import pandas as pd
```

```python
datos1 = pd.read_csv("./Datos_Ventas_Tienda.csv")
datos2 = pd.read_csv("./Datos_Ventas_Tienda2.csv")

df1 = pd.DataFrame(datos1)
df2 = pd.DataFrame(datos2)
df1
```

```python
df2
```

```python
df1.info()
```

```python
df2.info()
```

## Fusion de datos:

```python
df_ventas = pd.concat([df1,df2], ignore_index= True)
df_ventas
```

Ahora tenemos el DataFrame `df_ventas` con todos los datos no repetidos.

## Limpieza de datos:

```python
df_ventas["Fecha"] = pd.to_datetime(df_ventas["Fecha"])
df_ventas["Fecha"]
```

## Analisis de ventas:
#### Producto mas vendido:

```python
producto_estrella = df_ventas.groupby("Producto")["Cantidad"].sum()
producto_estrella.head(1)
```

#### Mes con mas ventas:

```python
meses = []
for f in df_ventas["Fecha"]:
    meses.append(f.month)
df_ventas["Meses"] = meses
df_ventas
```

```python
mes_estrella = df_ventas.groupby("Meses")["Total Venta"].sum().sort_values(ascending = False)
mes_estrella.head(1)
```

## Guardarlo en un .csv:

```python
df_ventas.to_csv("Resumen_ventas.csv")
```
