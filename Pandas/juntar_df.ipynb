```python
import pandas as pd
df1 = pd.DataFrame({"id":[1, 2, 3, 5, 6], "nombre": ["Antonio", "Mario", "Eugenia", "Celia", "Carlos"]})
df2 = pd.DataFrame({"id":[1, 2, 3, 4, 5], "suscripcion_activa": [True, False, True, True, False]})
df1
```

```python
df2
```

```python
df_unidas = pd.merge(df1, df2, on= "id")
df_unidas
```

## .merge
El atributo merge sirve para juntar DataFrames que compartan campo, el cual se indica con el campo `on`.<br> En caso de que un df tenga un valor y el otro no, no se guardará, es un `AND`.<br> Mas o menos es como una `foreign key`

```python
df_todo = pd.merge(df1, df2, on="id", how="outer")
df_todo
```

```python
# Creación del DataFrame df_a
df_a = pd.DataFrame({
    'id': [1, 2, 3],
    'Nombre': ['Ana', 'Beto', 'Carla']
})
df_a.set_index('id', inplace=True)

# Creación del DataFrame df_b
df_b = pd.DataFrame({
    'id': [1, 2, 3],
    'Edad': [25, 30, 35]
})
df_b.set_index('id', inplace=True)
df_a
```

```python
df_b
```

```python
df_combinado = df_a.join(df_b)
df_combinado
```

## .join
Tambien se pueden juntar los df con el atributo df, este se aplica sobre el df directamente y une uno principal con el secundario, es como si pusiesemos en el atributo merge el how = "left" puesto que da prioridad al primer df.<br>Algo a tener en cuenta de este atributo es que se basa no en los id, si no en los indices que tiene la tabla, por lo que habra que declararlos bien

```python
import pandas as pd
productos_df = pd.DataFrame({
    'ProductoID': [1, 2, 3, 4],
    'Nombre': ['Producto 1', 'Producto 2', 'Producto 3', 'Producto 4'],
    'Precio': [100, 150, 200, 250]
})

productos_df = productos_df.set_index('ProductoID')

categorias_df = pd.DataFrame({
    'CategoriaID': [1, 2, 3],
    'NombreCategoria': ['Categoría 1', 'Categoría 2', 'Categoría 3']
}).set_index('CategoriaID')

df_combinado = productos_df.join(categorias_df, how="right")
df_combinado
```
