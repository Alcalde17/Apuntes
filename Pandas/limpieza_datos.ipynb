# Limpieza de datos de un archivo

Primero que todo tenemos que iniciar los datos que queremos manipular

```python
import pandas as pd
datos = {"fruta": ["Limones", "Cocos", "Melocotones", "Albaricoques", "Sandias", "Fresas"],
                  "precio": [2.6, 2.1, 3.6, None, 4.9, 7.5],
                  "cantidad_vendida": [52, 21, 65, 29, None, 33]
                  }
df = pd.DataFrame(datos)
df
```

Como vemos, tenemos valores null en nuestras columnas, necesitamos limpiar esos datos de dos formas principalmente
1. Eliminarlos
2. Meter el promedio<br><br>Pero antes, para poder darnos cuenta si tenemos una gran cantidad de datos usaremos lo siguiente:

```python
df.info()
```

Esto nos puede mostrar los campos no nulos que hay por columnas y la cantidad de entradas que ha de haber,<br>
si estas no coinciden, quiere decir que hay campos nulos, por lo que debemos hacer limpieza.

### Rellenarlos con promedio

```python
valores_nuevos = {"precio":df["precio"].mean(), "cantidad_vendida": 0}
df_rellenados = df.fillna(valores_nuevos)
df_rellenados
```

```python
df_rellenados["cantidad_vendida"] = df_rellenados["cantidad_vendida"].astype(int)
df_rellenados
```

Se puede tambien limpiar datos duplicados, vamos a crear un nuevo df

```python
datos = {"id" : [1, 2, 2, 3, 4],
         "precio" : [2.3, 36.5, 36.5, 25.4, 5.5],
         "en_almacen" : [12, 6, 0, 5, 4]}
df_unicos = pd.DataFrame(datos)
df_unicos
```

Podemos ver como el id 2 es igual solo que tienen valores diferentes en almacén, por ello tendremos que especificar el campo que queremos que se fije para los duplicados:

```python
df_unicos = df_unicos.drop_duplicates(subset = "id")
df_unicos
```
